```python
# W6-HW-2

# Mission 7: 建立模擬的大學生學習資料
import numpy as np
import pandas as pd

# 隨機種子：同 seed 每次執行結果相同，方便課堂重現
rng = np.random.default_rng(42)

n = 60  # 學生人數

# 1. 學生編號
student_id = list(range(1, n + 1))

# 2. 科系 - 從三個科系隨機選擇
departments = ["資訊管理", "企業管理", "金融管理"]
dept_choices = rng.choice(departments, size=n)

# 3. 睡眠時數 - 常態分布 平均 6.7，標準差 1.1
sleep_hours = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕使用時數 - 常態分布 平均 5.0，標準差 1.5，限制 1~10
screen_time_hours = np.round(np.clip(rng.normal(loc=5.0, scale=1.5, size=n), 1.0, 10.0), 1)

# 5. 讀書時數 - 常態分布 平均 2.5，標準差 1.0，限制 0.5~6
study_hours = np.round(np.clip(rng.normal(loc=2.5, scale=1.0, size=n), 0.5, 6.0), 1)

# 6. 出席率 - 常態分布 平均 88，標準差 8，限制 60~100，整數
attendance_rate = np.round(np.clip(rng.normal(loc=88, scale=8, size=n), 60, 100)).astype(int)

# 組合成 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": dept_choices,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# ----------------------------------------------------
# 7. 計算小考成績 (根據原始圖表邏輯補齊)
# ----------------------------------------------------
base_score = 30  # 基礎分數
study_part = study_hours * 8           # 讀書時間帶來加分
screen_part = screen_time_hours * -2   # 螢幕時間帶來扣分
attendance_part = attendance_rate * 0.4 # 出席率的分數轉換
random_error = rng.normal(loc=0, scale=5, size=n) # 模擬不可控誤差

raw_scores = base_score + study_part + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error

# 限制分數在 0~100 之間，四捨五入並轉為 Numpy 的 int64 格式 (已修正 TypeError)
df["quiz_score"] = raw_scores.clip(0, 100).round().astype("int64")

# ----------------------------------------------------
# 8. 排序欄位並顯示完整的 60 筆資料
# ----------------------------------------------------
# 依照目標圖表定義要顯示的欄位與順序
selected_columns = [
    "student_id", "study_hours", "sleep_hours",
    "screen_time_hours", "attendance_rate", "quiz_score",
]

# 設定 Pandas 讓它最多能顯示 60 列 (這樣就不會被省略號 ... 蓋掉)
pd.set_option('display.max_rows', 60)

# 使用 display() 產生無折行的 HTML 表格 (適用於 Jupyter 筆記本環境)
display(df[selected_columns].head(60))
```
